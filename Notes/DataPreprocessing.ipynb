# Data Prepocessing
Data pre-processing is done before feeding the data to any ML model, the pre-processing involves the following steps: 
1. Importing the necessary libs
2. Reading the dataset
3. sanity check
4. Exploratory Data Analysis(EDA)
5. Missing Value Treatment
6. Outliers
7. Dublicates and garbage value treatment
8. Normalisation
9. Encoding of the data

## 1️⃣ Importing the Necessary Libraries

This step involves loading all the Python libraries or packages required for data handling, visualization, and processing.
Examples include:

pandas → for data manipulation

numpy → for numerical operations

matplotlib / seaborn → for visualization

scikit-learn → for preprocessing, encoding, scaling, and modeling

## 2️⃣ Reading the Dataset

This means loading your raw data (CSV, Excel, JSON, database, etc.) into a structured format — usually a DataFrame in pandas.

## 3️⃣ Sanity Check

A sanity check verifies that the data was loaded correctly and is in a usable state.
It includes:

Checking the number of rows and columns (df.shape)

Viewing data types (df.info())

Checking for null values and duplicates

Ensuring that the data makes sense (no impossible values like negative age)

✅ Purpose: To confirm the dataset’s correctness and detect immediate issues before analysis.

## 4️⃣ Exploratory Data Analysis (EDA)

EDA is the process of exploring and understanding the data using summary statistics and visualizations.
You examine:

- Distributions (mean, median, mode)

- Relationships between variables

- Correlations

- Patterns, trends, and anomalies
- Common tools: df.describe(), histograms, scatter plots, boxplots.

✅ Purpose: To gain insights and identify potential problems or patterns in the data.

## 5️⃣ Missing Value Treatment

Real-world data often has missing or null values. These need to be handled properly to avoid model errors.
Common methods:

- **Deletion**: Removing rows/columns with missing values

- **Imputation**: Filling missing values using mean, median, mode, or other logic

✅ Purpose: To ensure no gaps or NaN values cause issues during model training.

## 6️⃣ Outliers

Outliers are extreme or unusual values that differ greatly from other observations.
They can result from errors or rare conditions and may skew the results.
Detection methods:

- Boxplots

- Z-score

- Interquartile Range (IQR) method

✅ Purpose: To detect and treat abnormal data points for stable model performance.

## 7️⃣ Duplicates and Garbage Value Treatment

- Duplicates: Identical rows that can bias the data — detected with df.duplicated().

- Garbage values: Invalid or illogical entries (e.g., age = -5, salary = “??”).

✅ Purpose: To remove redundant or nonsensical information and keep the dataset consistent.

## 8️⃣ Normalization (or Scaling)

Normalization adjusts numerical values to a common scale without distorting their relationships.
Methods include:

- Min-Max Scaling: converts data to [0,1] range

- Standardization: centers data around mean 0 and standard deviation 1

✅ Purpose: To ensure all features contribute equally to the model and avoid bias due to varying scales.

# 9️⃣ Encoding of Data

Machine learning models work with numerical data, not text.
So, categorical (string) features must be converted into numbers using:

- Label Encoding: assigns each category an integer value

- One-Hot Encoding: creates binary columns (0/1) for each category

✅ Purpose: To transform categorical variables into numeric form for model compatibility.
